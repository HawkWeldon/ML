```python
import torch 
import torch.nn as nn
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
if torch.cuda.is_available(): print("GPU:", torch.cuda.get_device_name(0)) 
```

```python
class XORNN(nn.Module):
    
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.hidden = nn.Linear(2, 2)
        self.relu = nn.ReLU()
        self.output = nn.Linear(2, 1)

    def forward(self, x):
        z1 = self.hidden(x)
        x1 = self.relu(z1)
        z2 = self.output(x1)
        x2 = z2
        return x2
```

```python
model = XORNN().to(device)
model.load_state_dict(torch.load("XOR_model.pth", map_location=device))
```

```python
def predict_xor(a, b):
    x = torch.tensor([[float(a), float(b)]], device=device)
    model.eval()
    
    with torch.no_grad():
        logit = model(x)
        prob_coded = torch.sigmoid(logit)
        pred = (prob_coded >= 0.5).int()
        return pred.item()
```

```python
predict_xor(0,0)
```
